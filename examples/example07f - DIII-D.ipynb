# Example: Static inverse free-boundary equilibrium calculations (in DIII-D)

---

Here we generate a DIII-D equilibrium using the machine geometry and some discharge data in the [FUSE.jl repository](https://github.com/ProjectTorreyPines/FUSE.jl).

The equilibrium/profile parameters are a mix of data from the repository and **completely made up**; please experiment on your own and change them to more realistic values as you please!

### Import packages

```python
import matplotlib.pyplot as plt

import numpy as np
```

### Create the machine object

```python
# build machine

from freegsnke import build_machine

tokamak = build_machine.tokamak(
    active_coils_path="../machine_configs/DIII-D/DIII-D_active_coils.pickle",
    passive_coils_path="../machine_configs/DIII-D/DIII-D_passive_coils.pickle",
    limiter_path="../machine_configs/DIII-D/DIII-D_limiter.pickle",
    wall_path="../machine_configs/DIII-D/DIII-D_wall.pickle",
)
```

```python
# plot the machine

fig, ax = plt.subplots(figsize=(8, 10), dpi=80)

tokamak.plot(axis=ax, show=False)  # plots the active coils and passive structures
ax.plot(tokamak.limiter.R, tokamak.limiter.Z, "k-", linewidth=1.2)  # plots the limiter

ax.grid(alpha=0.5)
ax.set_aspect("equal")
ax.set_xlim(0.55, 2.85)
ax.set_ylim(-2.15, 2.15)
ax.set_xlabel(r"Major radius, $R$ [m]")
ax.set_ylabel(r"Height, $Z$ [m]")
```

### Instantiate an equilibrium

```python
from freegsnke import equilibrium_update

eq = equilibrium_update.Equilibrium(
    tokamak=tokamak,      # provide tokamak object
    Rmin=0.95, Rmax=2.43, # radial range
    Zmin=-1.42, Zmax=1.42,# vertical range
    nx=129,               # number of radial grid points (needs to be of the form 2**n + 1)
    ny=129,               # number of vertical grid points (needs to be of the form 2**n + 1)
)
```

### Load the static nonlinear solver

```python
from freegsnke import GSstaticsolver

GSStaticSolver = GSstaticsolver.NKGSsolver(eq, seed=42)
```

### Constraints

```python
# boundary and null constraints
Rx_upper, Zx_upper = 1.29, 1.12
Rx_lower, Zx_lower = 1.27, -1.17

null_points = [
    [Rx_lower, Rx_lower],
    [Zx_upper, Zx_lower],
]

isoflux_set = np.array([[[1.10, 2.23], [0.0, 0.0]]])

from freegsnke.inverse import Inverse_optimizer

constrain = Inverse_optimizer(
    null_points=null_points,
    isoflux_set=isoflux_set,
)
```

### The inverse solve

```python
GSStaticSolver.inverse_solve(
    eq=eq,
    profiles=profiles,
    constrain=constrain,
    target_relative_tolerance=5e-4,
    target_relative_psit_update=2e-3,
    max_solving_iterations=50,
    l2_reg=np.array([1e-14] * 24),
    verbose=True,
)
```

```python
# refine with a forward solve (now the currents are known)
GSStaticSolver.solve(eq=eq, 
                     profiles=profiles, 
                     constrain=None, 
                     target_relative_tolerance=1e-9,
                     verbose=False, # suppress output
                     )
```
